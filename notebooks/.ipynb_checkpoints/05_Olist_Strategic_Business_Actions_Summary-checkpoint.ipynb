````python
rfm = pd.read_csv(
    '../data/processed/rfm_customers.csv'
)

sales = pd.read_csv(
    '../data/processed/fact_sales.csv'
)
````

````python
lost_ids = rfm[
    rfm['Segment'] == 'Lost Customers'
]['customer_unique_id']

lost_sales = sales[
    sales['customer_unique_id'].isin(lost_ids)
]

lost_categories = (
    lost_sales
    .groupby('category')['payment_value']
    .sum()
    .sort_values(ascending=False)
    .head(20)
)

lost_categories
````

# Champions Product Profile

````python
champ_ids = rfm[
    rfm['Segment'] == 'Champions'
]['customer_unique_id']

champ_sales = sales[
    sales['customer_unique_id'].isin(champ_ids)
]

champ_categories = (
    champ_sales
    .groupby('category')['payment_value']
    .sum()
    .sort_values(ascending=False)
    .head(20)
)

champ_categories
````

# Comparing champions and losers

````python
champ_cat = (
    champ_sales
    .groupby('category')['payment_value']
    .sum()
)

lost_cat = (
    lost_sales
    .groupby('category')['payment_value']
    .sum()
)

compare = pd.concat(
    [champ_cat, lost_cat],
    axis=1
)

compare.columns = [
    'Champion_Revenue',
    'Lost_Revenue'
]

compare = compare.fillna(0)

compare['Loss_Ratio'] = (
    compare['Lost_Revenue']
    /
    (compare['Champion_Revenue'] + 1)
)

compare.sort_values(
    'Loss_Ratio',
    ascending=False
).head(20)
````

## Executive Insight: Revenue Leakage from Lost High-Value Customers

The analysis reveals a significant concentration of lost revenue within specific product categories, highlighting areas where customer retention failures have had the greatest financial impact.

### Key Findings

Several categories exhibit an exceptionally high **Lost-to-Champion Revenue Ratio**, indicating that a substantial portion of their historical value has already churned.

#### Critical Risk Categories

* **telefonia_fixa** generated approximately **$238K** from lost high-value customers compared to only **$11K** from active Champion customers.
* **cool_stuff** lost more than **$334K** in revenue while retaining only **$143K** from Champion customers.
* **agro_industria_e_comercio**, **moveis_sala**, and **brinquedos** also show significant revenue leakage, suggesting structural retention challenges.

These categories represent the highest opportunity for targeted retention and win-back initiatives.

### Category Collapse Indicators

Several categories demonstrate near-complete loss of valuable customer segments:

| Category           | Observation                 |
| ------------------ | --------------------------- |
| casa_conforto_2    | No Champion revenue remains |
| flores             | No Champion revenue remains |
| cds_dvds_musicais  | No Champion revenue remains |
| seguros_e_servicos | No Champion revenue remains |

In these categories, the business has effectively lost its highest-value customer base, indicating either declining market relevance, weak customer engagement, or ineffective retention strategies.

### Revenue Recovery Opportunity

The analysis suggests that customer churn is not evenly distributed across the business. Instead, a small number of categories account for a disproportionate share of lost revenue.

This finding aligns with the Pareto principle, where a limited set of product categories generates the majority of both business value and financial risk.

### Strategic Recommendations

1. **Prioritize Win-Back Campaigns**

   * Focus on categories with the highest Lost-to-Champion Revenue Ratios, particularly:

     * telefonia_fixa
     * cool_stuff
     * agro_industria_e_comercio
     * moveis_sala

2. **Investigate Category-Specific Churn Drivers**

   * Analyze pricing, delivery performance, customer reviews, and product quality within high-risk categories.

3. **Replicate Successful Retention Models**

   * Compare customer journeys from strong-performing categories against collapsed segments to identify retention best practices.

4. **Implement Early Warning Systems**

   * Use customer behavior patterns to identify high-value customers at risk before churn occurs.

### Business Impact

The results demonstrate that revenue growth is not solely dependent on acquiring new customers. A significant portion of potential growth can be achieved by preventing churn among existing high-value customers.

By addressing retention failures within a small number of critical categories, the company could unlock substantial incremental revenue without increasing acquisition costs.

# Executive Insights: Root Cause Analysis of Customer Churn

## 1. Key Finding: High Seller Concentration in Churn
* **Seller Concentration Index:** **0.301**
* **Strategic Takeaway:** Approximately **30% of all Lost Revenue** is driven by a very small, concentrated group of sellers. 
* **CEO Perspective:** The churn problem is *not* evenly distributed across the entire marketplace. The market as a whole is healthy, but the bleeding is heavily localized within a few specific high-risk sellers.

---

## 2. Hypothesis Validation: "Bad Product" vs. "Bad Seller"

### ❌ Hypothesis 1: Customer Churn is Driven by Product Categories (Rejected)
* **Observation:** High-churn categories like `relogios_presentes`, `informatica_acessorios`, and `moveis_decoracao` also appear prominently in the **Champions** segment.
* **Conclusion:** Since these categories simultaneously retain loyal customers and lose others, the category itself is not the inherent problem.

###  Hypothesis 2: Churn is Driven by the Seller Experience (Validated)
* **Observation:** There is a direct, near-perfect correlation between the highest-risk categories and the dominant sellers operating within them:
  * `telefonia_fixa` ➡️ Led by Seller `b37...`
  * `moveis_escritorio` ➡️ Led by Seller `7c67...`
  * `cool_stuff` ➡️ Led by Seller `7a67...`
* **Conclusion:** Customers do not leave because of *what* they buy (Category); they leave because of *who* they buy it from (Seller Experience). Churn is heavily tied to operational friction such as:
  * Product quality variance
  * Long delivery delays
  * Poor packaging or customer service

# Customer Loyalty & Repeat Purchase Rate Analysis by Category

## 1. Methodology & Logic
To understand which product categories successfully drive customer loyalty, we isolated customers with a purchase history of more than once (Repeat Customers) using RFM metrics. We then calculated the **Repeat Purchase Rate (RPR)** per category:

$$\text{Repeat Purchase Rate} = \left( \frac{\text{Unique Repeat Customers in Category}}{\text{Total Unique Customers in Category}} \right) \times 100$$

### Code Execution
```python
# 1. Identify loyal customers (Frequency > 1)
repeat_customers = rfm[rfm['Frequency'] > 1]['customer_unique_id']

# 2. Filter sales data for these repeat customers
repeat_sales = sales[sales['customer_unique_id'].isin(repeat_customers)]

# 3. Calculate unique repeat customers per category
repeat_category = repeat_sales.groupby('category')['customer_unique_id'].nunique()

# 4. Calculate the Return/Repeat Rate percentage
return_rate = (repeat_category / category_repeat) * 100
return_rate.sort_values(ascending=False)

# Strategic Business Actions for Management (C-Level Executive Summary)

Based on the derived **Repeat Purchase Rates (RPR)** combined with the previous **Seller Concentration Analysis**, management should immediately deploy the following three tactical playbooks:

---

### 1. The 3-Tier Marketing Budget Realignment

Management must stop treating all product categories equally and split the marketing budget into three distinct funnels:

* **The Retention Funnel (RPR > 10% | e.g., `la_cuisine`, `fashion`):** 
  * *Action:* Shift budget from paid acquisition (Google/Meta Ads) to **CRM and Loyalty Programs**. Implement personalized email flows, cross-selling, and subscription models. These customers want to return; we just need to nudge them.
* **The Acquisition Funnel (RPR 3% - 5% | e.g., `cool_stuff`, `electronics`):**
  * *Action:* Accept that these are high-volume, low-frequency goods. Management should focus strictly on **Optimizing CAC (Customer Acquisition Cost)** and conversion rate optimization (CRO) on the website. Do not waste budget on loyalty programs here.
* **The "One-and-Done" Funnel (RPR < 2% or `NaN` | e.g., `flores`, `seguros`):**
  * *Action:* **Blacklist** these categories from any retention or win-back campaigns. Treat them purely as high-margin, entry-level acquisition hooks.

---

### 2. Identifying and Plugging "Revenue Leakage"
* **The Conflict:** Categories like `telefonia_fixa` have a healthy baseline RPR of **5.53%** (meaning the product naturally triggers repeat purchases), yet they showed massive **Lost Revenue** in our previous churn report.
* **Management Decision:** This disparity is the smoking gun for **Bad Seller Operations**. The market wants the product, but specific high-concentration sellers are ruining the post-purchase experience (late deliveries, poor quality). Management must immediately audit and apply **Seller Penalties / Off-boarding** to save these high-potential categories.

---

### 3. Inventory & Supply Chain Priority
* **Action:** Categories with an RPR above 8% must be flagged as **Zero Stock-Out Priorities** in the warehouse. A loyal customer facing an "Out of Stock" status results in a permanent defection to competitors. 
* Low-RPR categories can be safely shifted to a decentralized **Drop-shipping** model to reduce Olist's warehouse overhead costs.

---

### 💡 Final Analytical Conclusion
By pivoting our strategy from **"fixing product categories"** to **"optimizing seller quality and aligning marketing spend with natural category RPRs"**, Olist can significantly reduce customer churn, lower overall CAC, and maximize the Lifetime Value (LTV) of our existing customer base without increasing ad spend.
