```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
rfm = pd.read_csv(
    '../data/processed/rfm_customers.csv'
)

sales = pd.read_csv(
    '../data/processed/fact_sales.csv'
)
```

```python
revenue_summary = pd.read_csv(
    '../data/processed/revenue_summary.csv'
)

champion_revenue = revenue_summary.loc[
    revenue_summary['Metric'] ==
    'Champion Revenue',
    'Value'
].iloc[0]

lost_revenue = revenue_summary.loc[
    revenue_summary['Metric'] ==
    'Lost Revenue',
    'Value'
].iloc[0]
```

```python
recovery_rates = [0.05, 0.10, 0.15, 0.20, 0.25]

results = []

for rate in recovery_rates:

    recovered_revenue = lost_revenue * rate

    new_total_revenue = (
        champion_revenue +
        recovered_revenue
    )

    growth_percent = (
        recovered_revenue /
        champion_revenue
    ) * 100

    results.append([
        f"{int(rate*100)}%",
        round(recovered_revenue, 2),
        round(new_total_revenue, 2),
        round(growth_percent, 2)
    ])

recovery_df = pd.DataFrame(
    results,
    columns=[
        'Retention Improvement',
        'Recovered Revenue',
        'New Revenue',
        'Revenue Growth %'
    ]
)

recovery_df
```

```python

plt.figure(figsize=(10,6))

plt.bar(
    recovery_df['Retention Improvement'],
    recovery_df['Recovered Revenue']
)

plt.title(
    'Potential Revenue Recovery from Churn Reduction'
)

plt.xlabel(
    'Retention Improvement'
)

plt.ylabel(
    'Recovered Revenue ($)'
)

plt.show()
```

```python
best_case = recovery_df.iloc[-1]

print(
f"""
If the company reduces churn among
high-value customers by 25%,

additional revenue generated:

${best_case['Recovered Revenue']:,.0f}

total revenue becomes:

${best_case['New Revenue']:,.0f}

equivalent to:

{best_case['Revenue Growth %']}% revenue growth
without increasing acquisition spend.
"""
)
```

# Executive Summary: Customer Retention Business Impact Analysis

## Objective

This analysis estimates the financial impact of reducing customer churn among high-value customers and quantifies the potential revenue recovery under different retention improvement scenarios.

---

## Key Findings

The simulation reveals a strong and direct relationship between customer retention and revenue growth.

| Retention Improvement | Recovered Revenue |
|----------------------|------------------:|
| 5% | $330K |
| 10% | $660K |
| 15% | $990K |
| 20% | $1.32M |
| 25% | $1.65M |

Even a modest **5% improvement in customer retention** generates approximately **$330,000** in additional revenue without increasing customer acquisition spending.

At the upper scenario, reducing churn by **25%** could recover more than **$1.65 million** in lost revenue.

---

## Strategic Insight

The analysis highlights a critical business reality:

> Revenue growth is not only driven by acquiring new customers. A significant portion of growth can be unlocked by preventing revenue leakage from existing high-value customers.

The simulation indicates that every 1% reduction in churn corresponds to approximately **$66,000 in recoverable revenue**, creating a highly attractive return on investment for retention-focused initiatives.

---

## Business Recommendations

### 1. Predict Churn Before It Happens

Implement predictive analytics models to identify high-value customers showing early signs of disengagement and intervene before they leave.

### 2. Prioritize High-Value Segments

Focus retention efforts on customer groups with the highest monetary value and greatest revenue contribution.

### 3. Shift Budget Allocation

Evaluate whether a portion of acquisition spending should be redirected toward:

- Customer Experience Improvement
- Loyalty Programs
- CRM Campaigns
- Post-Purchase Engagement
- Seller Quality Monitoring

### 4. Measure Revenue Leakage

Track churn not only as a customer metric but as a financial KPI directly linked to revenue recovery opportunities.

---

## Executive Conclusion

This analysis demonstrates that customer retention is one of the most powerful growth levers available to the business.

Recovering only a fraction of churned high-value customers can generate hundreds of thousands of dollars in incremental revenue while requiring significantly less investment than acquiring new customers.

The findings support a strategic shift from purely acquisition-driven growth toward a balanced model that combines acquisition, retention, and customer lifetime value optimization.
