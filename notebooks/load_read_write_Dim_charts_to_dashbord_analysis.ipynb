```python
import pandas as pd
from sqlalchemy import create_engine

```

```python

SERVER = "localhost"
DATABASE = "olistDW"

connection_string = (
    f"mssql+pyodbc://@{SERVER}/{DATABASE}"
    "?driver=ODBC+Driver+17+for+SQL+Server"
    "&trusted_connection=yes"
)

engine = create_engine(connection_string)

print("Connected Successfully")
```

```python
# Customer
dim_customer = pd.read_sql(
    "SELECT * FROM Dim_Customer",
    engine
)

# Product
dim_product = pd.read_sql(
    "SELECT * FROM Dim_Product",
    engine
)

# Seller
dim_seller = pd.read_sql(
    "SELECT * FROM Dim_Seller",
    engine
)

# Date
dim_date = pd.read_sql(
    "SELECT * FROM Dim_Date",
    engine
)
```

```python
dim_customer.to_csv(
    "../data/processed/Dim_Customer.csv",
    index=False
)

dim_product.to_csv(
    "../data/processed/Dim_Product.csv",
    index=False
)

dim_seller.to_csv(
    "../data/processed/Dim_Seller.csv",
    index=False
)

dim_date.to_csv(
    "../data/processed/Dim_Date.csv",
    index=False
)
```

```python

sales = pd.read_csv("../data/processed/fact_sales.csv")

rfm = pd.read_csv("../data/processed/rfm_customers.csv")
```

```python
lost_ids = rfm.loc[
    rfm["Segment"] == "Lost Customers",
    "customer_unique_id"
]

lost_sales = sales[
    sales["customer_unique_id"].isin(lost_ids)
]

lost_categories = (
    lost_sales
    .groupby("category", as_index=False)["payment_value"]
    .sum()
    .sort_values("payment_value", ascending=False)
)
```

```python
lost_categories.to_csv(
    "../data/processed/lost_categories.csv",
    index=False
)
```
